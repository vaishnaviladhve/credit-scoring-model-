````python
%%writefile README.md

# Credit Scoring Model

## Project Overview

This project develops a machine learning based Credit Scoring Model
to predict an individual's creditworthiness using financial information.

The project compares three classification algorithms:

1. Logistic Regression
2. Decision Tree
3. Random Forest

The models are evaluated using multiple classification metrics.

---

## Objective

The main objective of this project is to build a machine learning
classification model that can predict creditworthiness based on
financial and credit-related information.

The project also compares different machine learning algorithms
to understand their performance on the test dataset.

---

## Machine Learning Algorithms

The following machine learning algorithms are used:

### 1. Logistic Regression

Logistic Regression is a classification algorithm used to predict
the probability of a particular class.

It is used as a baseline classification model in this project.

### 2. Decision Tree

Decision Tree is a tree-based machine learning algorithm that makes
predictions using a sequence of decision rules.

### 3. Random Forest

Random Forest is an ensemble machine learning algorithm that combines
multiple decision trees to make predictions.

---

## Features

The model uses financial and credit-related features such as:

- Income
- Debt
- Credit Utilization
- Payment History
- Credit Accounts
- Loan Amount
- Late Payments
- Debt-to-Income Ratio
- Loan-to-Income Ratio
- Late Payment Risk

These features are used to identify patterns related to
creditworthiness.

---

## Feature Engineering

Additional financial features are created to improve the model,
including:

### Debt-to-Income Ratio

Debt-to-Income Ratio represents the relationship between a person's
debt and income.

### Loan-to-Income Ratio

Loan-to-Income Ratio represents the relationship between loan amount
and income.

### Late Payment Risk

Late Payment Risk is derived from late payment information and is
used as an additional feature for classification.

---

## Dataset

The dataset contains financial and credit-related information.

The dataset is divided into:

- Training dataset
- Testing dataset

The training data is used to train the machine learning models,
while the testing data is used to evaluate their performance.

---

## Data Preprocessing

The following preprocessing steps are performed:

1. Load the dataset.
2. Inspect the dataset.
3. Check missing values.
4. Handle missing values.
5. Separate features and target variable.
6. Encode categorical variables when required.
7. Scale numerical features when required.
8. Split the data into training and testing datasets.

---

## Train-Test Split

The dataset is divided into training and testing data.

The training data is used to train the models.

The testing data is used to evaluate how well the trained models
perform on unseen data.

---

## Model Training

Three classification models are trained:

```text
Logistic Regression
        ↓
Decision Tree
        ↓
Random Forest
````
